### Downloading Dataset From Google Drive Link

```python
import gdown
```

```python
url = "https://drive.google.com/file/d/1VMU6s_d0OaFdqu4OTubiUOL62hCxXW0z/view?usp=sharing"
```

```python
file_id = url.split("/")[-2]
print("Google Drive ID :", file_id)
```

```python
prefix = "http://drive.google.com/uc?/export=download&id="
gdown.download(prefix + file_id, "Kidney CT Scan File.zip")
```
